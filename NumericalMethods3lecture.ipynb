```python
def midd(a, b):
    return (a + b) / 2

def f1(x):
    return x**3 - x - 2

def main():
    a = 1.0
    b = 2.0
    E = 0.001

    while abs(f1(midd(a, b))) > E:
        if f1(a) * f1(midd(a, b)) < 0:
            b = midd(a, b)
        else:
            a = midd(a, b)

    print( midd(a, b), f1(midd(a, b)) )

if __name__ == "__main__":
    main()
```

```python
def fx(x):
    return x**3 - x - 2

def main():
    x0 = 1.0
    x1 = 2.0
    E = 0.001

    while True:
        x_next = x1 - (fx(x1) * (x1 - x0)) / (fx(x1) - fx(x0))
        
        if abs(x_next - x1) <= E:
            x0 = x1
            x1 = x_next
            break
            
        x0 = x1
        x1 = x_next

    print(f"{x1:.6f}")
    print(f"{abs(x1 - x0):.6f}")

if __name__ == "__main__":
    main()
```

```python
def fn(x):
    return x**3 - x - 2

def fnd(x):
    return 3*x**2 - 1


def main():
    a = 1.0
    E = 0.001
    x = a
    while True:
        x1 = x - fn(x) / fnd(x)
        if abs(x1 - x) <= E:
            break
        else:
            x = x1
    print(x1, f"{abs(fn(x1)):.10f}")

if __name__ == "__main__":
    main()
    
```
