```python
import speech_recognition as sr

def recognize_from_microphone():
    """Распознавание речи с микрофона"""
    recognizer = sr.Recognizer()
    
    with sr.Microphone() as source:
        print("Адаптация под фоновый шум...")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        
        print("Запись в течение 4 секунд (говорите)...")
        audio_data = recognizer.record(source, duration=4)
        print("Запись завершена.")
        
        print("Распознавание текста...")
        try:
            # Для распознавания на русском языке используйте language="ru-RU"
            # Для английского языка используйте language="en-US"
            text = recognizer.recognize_google(audio_data, language="ru-RU")
            print(f"Распознанный текст: {text}")
        except sr.UnknownValueError:
            print("Не удалось распознать речь.")
        except sr.RequestError as e:
            print(f"Ошибка сервиса распознавания: {e}")

def recognize_from_file(file_path):
    """Распознавание речи из аудиофайла (WAV)"""
    recognizer = sr.Recognizer()
    
    try:
        with sr.AudioFile(file_path) as source:
            print(f"Загрузка аудиофайла: {file_path}")
            audio_data = recognizer.record(source)
            
            print("Распознавание текста...")
            text = recognizer.recognize_google(audio_data, language="ru-RU")
            print(f"Распознанный текст: {text}")
    except FileNotFoundError:
        print(f"Файл '{file_path}' не найден.")
    except sr.UnknownValueError:
        print("Не удалось распознать речь в файле.")
    except sr.RequestError as e:
        print(f"Ошибка сервиса распознавания: {e}")

if __name__ == "__main__":
    # Запуск распознавания с микрофона
    recognize_from_microphone()
    
    # Чтобы распознать файл, раскомментируйте строчку ниже и укажите путь к аудиофайлу (.wav):
    # recognize_from_file("your_audio_file.wav")
```
