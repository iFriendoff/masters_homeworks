```python
object_cols = handlemse.select_dtypes(include=['object']).columns
handlemse = handlemse.drop(columns=object_cols)
features = [
    'odometer_km',             # Пробег
    'cycle_count',            # Количество циклов
    'cell_temperature_max',   # Макс. температура
    'internal_resistance'     # Внутреннее сопротивление
]
target = 'battery_health_percent'  # Твой y (Здоровье батареи в %)

handlemse = handlemse[features + [target]].copy()
```

```python
print(handlemse.isna().sum())

handlemse = handlemse.dropna()
```
