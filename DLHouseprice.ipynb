```python
import tensorflow as tf
import pandas as pd
df = pd.read_csv(r"house_prices.csv")
```

```python
print(df.info())
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(df.shape[1] - 1,)),
    tf.keras.layers.Dense(32, activation='relu'),       
    tf.keras.layers.Dense(1)  
])
import keras
from sklearn.model_selection import train_test_split
X = df.drop('Price (in rupees)', axis=1)
Y = df['Price (in rupees)']


X_train, X_val, y_train, y_val = train_test_split(X, Y, test_size=0.2, random_state=42)

model.compile(optimizer='adam', loss=keras.losses.MeanSquaredError(), metrics=['accuracy', tf.keras.metrics.Precision()])
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_val, y_val))
precision = model.precision_score(y_val, model.predict(X_val).round())
accuracy = model.evaluate(X_val, y_val)[1]
recall = model.recall_score(y_val, model.predict(X_val).round())
f1_score = 2 * (precision * recall) / (precision + recall)
```
