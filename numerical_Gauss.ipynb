```python
a = [[10,-7,0.0],
     [-3,2,6],
     [5,-1,5]]

b = [7,4,6]
num = 0
for i in range(len(a)):
    if a[i][i] == 0:
        num+= 1
        a[i], a[i+1] = a[i+1], a[i]
        b[i], b[i+1] = b[i+1], b[i]

    for j in range(i + 1, len(a)):
        с = a[j][i] / a[i][i]

        for k in range(len(a[i])):
            a[j][k] = a[j][k] - с * a[i][k]

        b[j] = b[j] - с * b[i]

print(a)
print(b)
for i in range(len(a) - 1, -1, -1):
    x = b[i] / a[i][i]
    for j in range(i - 1, -1, -1):
        b[j] = b[j] - a[j][i] * x
    print("x{} = {}".format(i + 1, x))

for i in range (len(a)):
    determinant = (-1) ** num * a[i][i]
print("Determinant = {}".format(determinant))
```
